```python
import pandas as pd
import numpy as np
import os

# 1. File paths define karein
INPUT_FILE = r"C:\Users\dell\Downloads\train.csv" 
OUTPUT_FILE = r"C:\Users\dell\Downloads\cleaned_data.csv"      
LOG_FILE = r"C:\Users\dell\Downloads\cleaning_log.txt"          

# 2. Data load karein
if os.path.exists(INPUT_FILE):
    df = pd.read_csv(INPUT_FILE)
    print(f" Data loaded! Shape: {df.shape}")
    
    # Cleaning ki history track karne ke liye ek list
    cleaning_history = []
    cleaning_history.append(f"Initial Shape: {df.shape}")
else:
    print(" Error: 'train.csv' file nahi mili. Path check karein.")

# 3. Data ka ek preview dekhein
df.head()
```

```python

df.columns = (
    df.columns.str.lower()                  
    .str.strip()                            
    .str.replace(" ", "_")                   
    .str.replace(r"[^a-z0-9_]", "", regex=True) 
)

cleaning_history.append("Step 2: Column names standardized to snake_case.")

print("Naye Column Names:")
print(df.columns.tolist())
```

```python
# 1. Count duplicate rows
duplicate_count = df.duplicated().sum()

# 2. Remove duplicates if they exist
if duplicate_count > 0:
    df.drop_duplicates(inplace=True)
    cleaning_history.append(f"Step 3: Removed {duplicate_count} duplicate rows.")
    print(f" Success: {duplicate_count} duplicate rows have been deleted.")
else:
    cleaning_history.append("Step 3: No duplicate rows found.")
    print(" Success: No duplicate rows were found in the dataset.")

# 3. Check current dataset size
print(f"Current Data Shape: {df.shape}")
```

```python
# 1. Loop through columns to find missing values
for col in df.columns:
    missing_count = df[col].isnull().sum()
    
    if missing_count > 0:
        # Check if the column is numeric (float or int)
        if df[col].dtype in ["float64", "int64"]:
            # Fill with Mean
            column_mean = df[col].mean()
            df[col] = df[col].fillna(column_mean)
            cleaning_history.append(f"Step 4: Filled {missing_count} nulls in '{col}' using Mean.")
        else:
            # Fill with Mode (Categorical/Object)
            column_mode = df[col].mode()[0]
            df[col] = df[col].fillna(column_mode)
            cleaning_history.append(f"Step 4: Filled {missing_count} nulls in '{col}' using Mode.")

print(" Success: All missing values have been handled.")
```

```python
# 1. Save the cleaned DataFrame to a CSV file
df.to_csv(OUTPUT_FILE, index=False)

# 2. Save the cleaning history to a text log file
with open(LOG_FILE, "w") as f:
    f.write("--- DATA CLEANING REPORT ---\n")
    for line in cleaning_history:
        f.write(line + "\n")
    f.write("--- End of Report ---")

# 3. Final summary display
print("========================================")
print(" DATA CLEANING COMPLETED SUCCESSFULLY!")
print(f" Final Data Shape: {df.shape}")
print(f"Cleaned File Saved As: {OUTPUT_FILE}")
print(f" Audit Log Saved As: {LOG_FILE}")
print("========================================")
```
